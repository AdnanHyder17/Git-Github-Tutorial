## 1.1 What is Git?

**Git is a free and open-source distributed version control system (DVCS)**. It records changes to files so that you can review earlier versions, understand how a project evolved, and collaborate with other developers.

Git is commonly used for source code, but it can track many kinds of text-based files, including documentation, configuration files, and notebooks.

### What Git helps you do

- Keep a history of changes.
- Compare versions of files.
- Restore earlier work when necessary.
- Develop features on separate branches.
- Combine work from multiple contributors.
- Identify which commit introduced a change.

Git does not automatically decide which code is correct, prevent every conflict, or host your project online. It gives you tools to manage changes safely and deliberately.

## 1.2 What is version control?

**Version control** is the practice of recording changes to files over time.

Without version control, people often create copies such as `project-final`, `project-final-v2`, and `project-final-really-final`. This quickly becomes difficult to manage.

With Git, you maintain one working project and create commits: saved checkpoints with messages describing the changes.

## 1.3 What is a repository?

A **repository (repo)** is a project directory whose history is tracked by Git. Git stores repository metadata in a hidden `.git` directory.

A repository contains:
- Your project files.
- Git's internal database and metadata.
- The history of commits, branches, and other references.

### Local repository

A local repository is the copy stored on your computer. You can edit files, create commits, inspect history, and work offline.

### Remote repository

A remote repository is another copy of a Git repository hosted somewhere accessible over a network. In this course, we use **GitHub** as the hosting platform.

A remote is not necessarily GitHub; it is simply a named reference to another repository location.

## 1.4 Git vs GitHub

| Git | GitHub |
|---|---|
| Version control software | Online platform for hosting Git repositories |
| Runs locally on your computer | Accessed through the web and Git tools |
| Tracks commits, branches, and file changes | Adds collaboration features such as pull requests, issues, and reviews |
| Can work without an internet connection | Network access is needed for hosted collaboration |

Git is the tool; GitHub is one service built around Git.

## 1.5 How Git supports collaboration

A typical team workflow looks like this:

1. A developer clones the project from GitHub.
2. They create a branch for a task.
3. They edit files and commit their work locally.
4. They push the branch to GitHub.
5. They open a pull request (PR) for discussion and review.
6. After review, the changes are merged into the shared branch.
7. Other developers fetch or pull the updated work.

Each developer has a local repository and can work independently. Git helps combine the work and highlights conflicting edits that need human resolution.

## 1.6 Install Git

```bash

# Fedora
sudo dnf install git

xcode-select --install
```

On Windows, follow the Git for Windows installer. The default options are suitable for most beginners. Git Bash is a useful terminal, while PowerShell also works with Git.

## 1.8 Which terminal should you use?

Git commands are mostly the same in PowerShell, Command Prompt, Git Bash, macOS Terminal, and Linux shells.

Some surrounding shell commands differ. For example, listing files uses `dir` in PowerShell/CMD and `ls` in Git Bash/macOS/Linux. This tutorial labels shell-specific examples when needed.
