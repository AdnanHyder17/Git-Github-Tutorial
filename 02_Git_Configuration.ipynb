# 2. Git Configuration

Configure Git identity and understand configuration scopes before creating commits. The examples work across Windows, macOS, and Linux.

## 2.1 Why configure Git?

Git records an author name and email address in each commit. Configure these values so your commits are attributed correctly.

The values are labels stored in commit metadata; they are not, by themselves, proof of identity. If you use GitHub, configure an email address that GitHub can associate with your account, or use a GitHub-provided noreply address.

## 2.2 Set your global identity

`--global` applies the setting to your user account on this computer, for all repositories unless overridden locally.

```bash
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
```

Replace the example values with your own preferred commit identity.

## 2.3 Set repository-specific identity

Inside a particular repository, you can set a local identity:

```bash
git config user.name "Work Name"
git config user.email "work@example.com"
```

Local configuration overrides the global value for that repository.

## 2.4 Configuration scopes

| Scope | Command option | Where it applies |
|---|---|---|
| System | `--system` | All users on the computer |
| Global | `--global` | Current operating-system user |
| Local | `--local` | Current repository |
| Worktree | `--worktree` | Current worktree, when enabled |

The usual precedence is worktree, local, global, then system. Most learners only need global and local configuration.

## 2.5 View configuration

```bash
# Show all effective configuration values
git config --list

# Show a particular value
git config user.name
git config user.email
```

## 2.6 Set the default branch name

Many projects use `main` as their default initial branch:

```bash
git config --global init.defaultBranch main
```

This affects newly initialized repositories. It does not rename branches in existing repositories.

## 2.7 Configure a default editor (optional)

You can configure the editor Git opens for commit messages and other tasks. Examples:

```bash
# Visual Studio Code (if the `code` command is available)
git config --global core.editor "code --wait"

# Nano
git config --global core.editor "nano"
```

On Windows, the exact editor command depends on how the editor was installed and whether it is available on PATH.

## 2.8 Remove or change a setting

```bash
# Change a value
git config --global user.name "New Name"

# Remove a global value
git config --global --unset user.name
```

Be careful when removing configuration: a repository may rely on a global value.

## 2.9 Configuration files

Git commonly reads configuration from:
- System config.
- User/global config.
- Repository-local `.git/config`.
- Worktree-specific config, when enabled.

You usually should change settings with `git config` rather than editing these files manually.

## Key takeaways

- Configure `user.name` and `user.email` before making commits.
- `--global` is convenient for your usual identity.
- Local settings override global settings for one repository.
- `git config --list` and `--show-origin` help diagnose configuration.
