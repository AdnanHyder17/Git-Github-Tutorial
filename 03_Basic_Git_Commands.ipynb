## 3. Basic Git Commands

Learn the everyday commands used to create a repository, inspect its state, and navigate Git's help system.

### 3.1 Open a terminal and choose a project folder

Navigate to your project directory using `cd`:

```bash
cd path/to/your/projects
```

**Alternative (Windows – Git Bash):** Open your project folder, right-click an empty area, and select **Open Git Bash here**. Git Bash will open directly in that directory.

**Shell note:**
- PowerShell and Command Prompt support `cd`.
- Git Bash, macOS, and Linux terminals also support `cd`.
- Quote paths containing spaces, e.g., `cd "C:/Users/Your Name/Projects"` in Git Bash or `cd "C:\Users\Your Name\Projects"` in PowerShell.


### 3.2 Get help

```bash
git help
git help status
git status --help
```

Depending on your setup, help may open in a browser, a terminal pager, or a manual viewer. Press `q` to exit a pager.

You can also use concise help:

```bash
git status -h
```

### 3.3 Initialize a repository

Create or enter a project directory, then run:

```bash
git init
```

This creates a `.git` directory and turns the current directory into a Git repository. It does not automatically add or commit your files.

To initialize a new directory:

```bash
mkdir git-practice
cd git-practice
git init
```

`mkdir` and path syntax can vary slightly by shell, but the Git command is the same.

### 3.4 Check repository status

```bash
git status
```

This reports the current branch and whether files are untracked, modified, staged, or clean.

### 3.5 Understand the `.git` directory

The `.git` directory contains Git's repository database and configuration. Do not manually delete or casually edit its contents. If it is removed, the project files remain, but the repository history and metadata are lost.


### 3.6 A safe first practice sequence

```bash
mkdir git-basics
cd git-basics
git init
git status
```

If the folder already exists, use it rather than creating it again.

### Key takeaways

- `git init` starts tracking a directory as a repository.
- `git status` is one of the most useful commands; run it often.
- `git help <command>` explains commands.
