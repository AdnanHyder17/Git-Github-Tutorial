# 4. Working with Repositories

Understand the difference between creating a repository locally and obtaining an existing repository from GitHub.

> **Environment note:** Concepts apply to Windows, macOS, and Linux. Commands are shown in a cross-platform way wherever possible. Screenshot placeholders are intentionally written for **Windows only** (PowerShell, Command Prompt, or Git Bash).

## 4.1 Create a local repository

A local repository starts on your computer. Navigate into the project directory and run:

```bash
git init
```

Git creates the `.git` metadata directory. Your project files are not automatically committed.

## 4.2 Create a repository on GitHub

1. Sign in to [GitHub](https://github.com/).
2. Select **New repository**.
3. Enter a repository name and optional description.
4. Choose Public or Private.
5. For a first local-to-GitHub practice, consider creating an empty repository without a README, license, or `.gitignore`. This avoids an unrelated initial-history merge.
6. Create the repository.

GitHub will display a repository URL. It commonly looks like:

```text
https://github.com/USERNAME/REPOSITORY.git
```

Use your own URL; do not copy the example literally.

## 4.3 Clone an existing repository

**Cloning** downloads a repository and its available history to your computer, and configures a remote (usually named `origin`).

```bash
git clone https://github.com/USERNAME/REPOSITORY.git
```

To choose a different local folder name:

```bash
git clone https://github.com/USERNAME/REPOSITORY.git my-project
```

Then enter the folder:

```bash
cd my-project
git status
```

## 4.4 Clone using SSH (optional)

If you have configured an SSH key with GitHub, you can clone using the SSH URL:

```bash
git clone git@github.com:USERNAME/REPOSITORY.git
```

HTTPS and SSH are both valid. SSH requires key setup; HTTPS may prompt for browser-based authentication or a credential manager. GitHub account passwords are not accepted for Git HTTPS authentication; use the supported sign-in flow or a personal access token where appropriate.

## 4.5 Local vs remote: the mental model

- **Working directory:** files you are currently editing.
- **Local repository:** commits and history stored on your computer.
- **Remote repository:** another repository, often hosted on GitHub.

Cloning creates a local copy of a remote repository. After that, local work and remote work are synchronized using commands such as `fetch`, `pull`, and `push`.

## 4.6 Inspect the repository

```bash
git status
git branch
git remote -v
```

- `git status` shows the working state.
- `git branch` lists local branches.
- `git remote -v` lists remote names and URLs.

## 4.7 Common beginner mistakes

- Running `git init` in the wrong folder.
- Cloning into a directory that already contains unrelated files.
- Confusing the GitHub webpage with the local working directory.
- Assuming that a local commit automatically appears on GitHub. It does not; you must push it.

## Key takeaways

- Use `git init` to start a new local repository.
- Use `git clone` to copy an existing repository.
- A clone normally configures `origin` automatically.
- Local commits remain local until pushed.
