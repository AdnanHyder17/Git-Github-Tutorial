# 5. Staging and Committing

Learn Git's central workflow: edit files, stage selected changes, and save them as commits.

> **Environment note:** Concepts apply to Windows, macOS, and Linux. Commands are shown in a cross-platform way wherever possible. Screenshot placeholders are intentionally written for **Windows only** (PowerShell, Command Prompt, or Git Bash).

## 5.1 The three areas of Git

A useful Git mental model has three areas:

1. **Working directory:** the files you currently edit.
2. **Staging area (index):** the changes selected for the next commit.
3. **Repository history:** committed snapshots.

A commit records the staged content, not every change currently present in your folder.

## 5.2 Check the current state

```bash
git status
```

Git identifies untracked files and modifications. An untracked file is present in your folder but is not yet tracked by Git.

## 5.3 Stage a file

```bash
git add README.md
```

This stages the current contents of `README.md`.

Stage all changes in the current directory:

```bash
git add .
```

`git add -A` stages additions, modifications, and deletions throughout the working tree. Review `git status` before committing, especially when using broad staging commands.

## 5.4 Unstage a file

If you staged a file by mistake:

```bash
git restore --staged README.md
```

This removes it from the staging area but keeps your working-directory changes.

Older Git versions may use:

```bash
git reset HEAD README.md
```

## 5.5 Create a commit

```bash
git commit -m "Add project introduction"
```

A commit is a saved checkpoint of the staged changes. A good commit message briefly describes the change.

Examples:
- `Add login form validation`
- `Fix date parsing issue`
- `Update installation instructions`

Avoid vague messages such as `changes`, `final`, or `stuff`.

## 5.6 Inspect commit history

```bash
git log
git log --oneline
git log --oneline --graph --decorate --all
```

The compact form is useful for a quick overview. Press `q` to exit the pager if one opens.

## 5.7 Compare changes with diff

```bash
# Changes not yet staged
git diff

# Changes staged for the next commit
git diff --staged

# Summary of changed files
git diff --stat
```

`git diff` helps you review exactly what you are about to save.

## 5.8 Modify a file after staging

If you stage a file and then edit it again, the new edits are not automatically staged. Run `git add` again to include the latest changes.

## 5.9 A complete example workflow

```bash
git status
git add README.md
git status
git diff --staged
git commit -m "Add README"
git log --oneline
```

## 5.10 What makes a good commit?

- Keep commits focused on one logical change.
- Use clear, action-oriented messages.
- Review staged changes before committing.
- Avoid committing passwords, API keys, private keys, or other secrets.
- Commit often enough that each checkpoint is meaningful.

## Key takeaways

- `git add` stages changes.
- `git commit` saves staged changes to local history.
- `git status` shows what is staged and unstaged.
- `git diff` reviews changes before committing.
- A commit is local until it is pushed to a remote.
