# 6. Understanding Branches

Learn how branches let you develop features or fixes independently without making every experiment directly on the main line of development.

> **Environment note:** Concepts apply to Windows, macOS, and Linux. Commands are shown in a cross-platform way wherever possible. Screenshot placeholders are intentionally written for **Windows only** (PowerShell, Command Prompt, or Git Bash).

## 6.1 What is a branch?

A branch is a movable reference to a commit. It gives you an independent line of development.

Most new repositories use `main` as the primary branch. A developer might create a branch named `feature/login` to work on a login feature without changing `main` until the work is ready.

Branches are lightweight; creating one does not duplicate the entire project folder in the way manually copying a directory would.

## 6.2 List branches

```bash
git branch
git branch --all
```

The asterisk marks the currently checked-out branch.

## 6.3 Create a branch

```bash
git branch feature/login
```

This creates the branch but does not switch to it.

## 6.4 Switch branches

```bash
git switch feature/login
```

Create and switch in one command:

```bash
git switch -c feature/profile
```

Older equivalent commands are:

```bash
git checkout feature/login
git checkout -b feature/profile
```

For new learners, prefer `git switch` for branch navigation and `git restore` for restoring files; these commands make intent clearer.

## 6.5 Rename a branch

Rename the current branch:

```bash
git branch -m new-branch-name
```

Rename a different local branch:

```bash
git branch -m old-name new-name
```

## 6.6 Delete a branch

Delete a branch that Git considers safely merged:

```bash
git branch -d feature/login
```

Force-delete a local branch (this can discard unmerged work):

```bash
git branch -D feature/login
```

Do not force-delete unless you understand what work may be lost.

## 6.7 Branching workflow

A simple feature workflow:

```bash
git switch main
git pull
git switch -c feature/search
# Make changes, then stage and commit them.
git add .
git commit -m "Add search feature"
```

The `git pull` step updates the local `main` from its configured remote. It assumes the repository has a remote and an upstream branch configured.

## 6.8 Important branch concepts

- Branches are references to commits, not separate projects.
- Switching branches changes the files in your working directory to match the target branch.
- Commit or stash work before switching if it might conflict with the target branch.
- A local branch and a remote-tracking branch are different references.

## Key takeaways

- Use branches to isolate work.
- `git switch -c name` creates and switches to a branch.
- `git branch` lists local branches.
- Delete merged branches with `git branch -d`.
