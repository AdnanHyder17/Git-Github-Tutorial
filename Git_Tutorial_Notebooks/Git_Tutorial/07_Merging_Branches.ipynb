Understand how Git combines work from different branches, what fast-forward and three-way merges mean, and how to resolve merge conflicts.

## 7.1 What is merging?

Merging combines the changes from one branch into another. Usually, you switch to the branch that should receive the changes, then merge the source branch into it.

Example: merge `feature/search` into `main`:

```bash
git switch main
git merge feature/search
```

The destination branch is the branch you have checked out.

## 7.2 Fast-forward merge

If the destination branch has not advanced since the feature branch was created, Git may simply move the destination branch pointer forward. This is a **fast-forward merge**.

No merge commit is required in this case.

## 7.3 Three-way merge

If both branches have new commits since they diverged, Git uses their common ancestor and the two branch tips to combine changes. This often creates a merge commit.

```bash
git switch main
git merge feature/search
```

Git may automatically complete the merge if it can combine the changes safely.

## 7.4 What is a merge conflict?

A merge conflict occurs when Git cannot automatically determine how to combine changes. A common example is when two branches edit overlapping lines in the same file.

Git marks the conflicted section in the file:

```text
<<<<<<< HEAD
Content from the current branch
=======
Content from the branch being merged
>>>>>>> feature/example
```

These marker lines are not valid final content. Edit the file and choose the correct combined result, then remove all conflict markers.

## 7.5 Resolve a conflict

1. Run `git status` to identify conflicted files.
2. Open each conflicted file in an editor.
3. Decide which content should remain (you may combine both versions).
4. Remove the conflict markers.
5. Save the file.
6. Stage the resolved file with `git add`.
7. Complete the merge with `git commit` if Git has not already completed it.

Example:

```bash
git status
git add path/to/conflicted-file
git commit
```

For a merge in progress, `git commit` opens the merge commit message if one is needed.

## 7.6 Abort a merge

If you want to stop a merge that has not been completed:

```bash
git merge --abort
```

This attempts to return to the state before the merge began. Avoid starting a merge with unrelated uncommitted changes in your working directory.

## 7.7 Inspect merge history

```bash
git log --oneline --graph --decorate --all
```

This graph view helps you understand how branches diverged and came together.

## 7.8 Merge vs rebase (conceptual distinction)

- **Merge** combines histories and preserves the branch structure; it may create a merge commit.
- **Rebase** replays commits onto a new base, creating new commit IDs.

Rebase can produce a linear history, but rewriting published/shared history can disrupt collaborators. For shared branches, coordinate before rebasing or force-pushing.

## Key takeaways

- Switch to the destination branch before merging.
- Git may perform a fast-forward or a three-way merge.
- Conflicts require a person to choose the correct final content.
- Use `git status` to locate conflicts.
- `git merge --abort` can cancel an in-progress merge.
