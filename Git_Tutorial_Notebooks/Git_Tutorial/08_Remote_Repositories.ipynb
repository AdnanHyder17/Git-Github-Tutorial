# 8. Remote Repositories

Learn how your local repository communicates with a remote GitHub repository using remotes, fetch, pull, and push.

> **Environment note:** Concepts apply to Windows, macOS, and Linux. Commands are shown in a cross-platform way wherever possible. Screenshot placeholders are intentionally written for **Windows only** (PowerShell, Command Prompt, or Git Bash).

## 8.1 What is a remote?

A remote is a named reference to another Git repository. A clone normally creates a remote named `origin`.

Inspect configured remotes:

```bash
git remote
git remote -v
```

`-v` shows the fetch and push URLs.

## 8.2 Add a remote

If you initialized a repository locally and created an empty repository on GitHub, connect them:

```bash
git remote add origin https://github.com/USERNAME/REPOSITORY.git
```

Replace the URL with your own repository URL.

Check it:

```bash
git remote -v
```

## 8.3 Push local commits to GitHub

Push your current branch and set its upstream:

```bash
git push -u origin main
```

The `-u` option sets the upstream tracking relationship. On later pushes from that branch, you can often use:

```bash
git push
```

If your current branch is not named `main`, use its actual name.

## 8.4 Fetch: download remote information

```bash
git fetch origin
```

Fetch downloads commits and updates remote-tracking references, but does not merge those changes into your current branch or change your working files.

This is useful when you want to inspect remote updates before integrating them.

## 8.5 Pull: fetch and integrate

```bash
git pull
```

`git pull` fetches and then integrates changes from the configured upstream branch. The integration is commonly a merge; configuration can also make pull rebase.

To make the strategy explicit:

```bash
git pull --no-rebase
git pull --rebase
```

Choose a consistent team workflow. If you are unsure, inspect `git status` and your branch tracking setup before pulling.

## 8.6 Understand origin and upstream

- `origin` is a conventional name for the remote created by `git clone` or added manually.
- `upstream` is often used as the name of the original project repository when working from a fork.
- These are names, not special Git keywords; teams can choose other names.

## 8.7 Inspect tracking relationships

```bash
git branch -vv
git status
```

`git branch -vv` shows local branches and their upstream tracking branches, when configured.

## 8.8 Change or remove a remote

```bash
git remote set-url origin https://github.com/USERNAME/NEW-REPOSITORY.git
git remote remove origin
```

Removing a remote does not delete the GitHub repository or your local commits.

## 8.9 Authentication

For GitHub over HTTPS, use the supported browser sign-in / credential manager flow or a personal access token when required. GitHub does not accept your account password for Git HTTPS operations.

For SSH, create an SSH key pair, add the public key to your GitHub account, and use the SSH repository URL. Never share your private key or token.

## 8.10 A common local-to-GitHub sequence

```bash
git status
git add .
git commit -m "Save project changes"
git push -u origin main
```

Before pushing, make sure the intended branch is checked out and the commit does not contain secrets.

## Key takeaways

- A remote is a named connection to another repository.
- `origin` is the common default remote name.
- `fetch` downloads remote history without integrating it into your current branch.
- `pull` fetches and integrates.
- `push` sends your local commits to the remote.
