## 9.1 Collaboration overview

GitHub provides collaboration features around Git repositories. A common team workflow is:

1. Obtain the project (clone it or fork it).
2. Create a branch for a task.
3. Make focused commits.
4. Push the branch to GitHub.
5. Open a pull request.
6. Discuss and review the proposed changes.
7. Update the branch if requested.
8. Merge the pull request.
9. Synchronize your local repository.

## 9.2 What is a pull request?

A **pull request (PR)** is a GitHub feature for proposing changes from one branch into another. It provides a place to:
- Review code differences.
- Discuss implementation choices.
- Leave comments on specific lines.
- Run automated checks.
- Approve or request changes.
- Merge the work when it is ready.

A PR is not itself a Git command; it is a GitHub collaboration feature.

## 9.3 Team workflow when you have write access

Start from an up-to-date `main` branch:

```bash
git switch main
git pull
git switch -c feature/update-docs
```

Make your changes, then:

```bash
git status
git add .
git commit -m "Update project documentation"
git push -u origin feature/update-docs
```

On GitHub, open a pull request:
- Base: `main` (the branch receiving the changes).
- Compare: `feature/update-docs` (the branch containing your changes).
- Add a clear title and description.
- Explain what changed and how it was tested.

## 9.4 Fork-based workflow

A **fork** is your GitHub copy of another user's repository. It is commonly used when you do not have write access to the original project.

Typical process:

1. Click **Fork** on the original GitHub repository.
2. Clone your fork.
3. Add the original repository as a remote named `upstream`.
4. Create a feature branch.
5. Commit and push to your fork (`origin`).
6. Open a PR from your fork's branch into the original repository.

Example:

```bash
git clone https://github.com/YOUR-USERNAME/PROJECT.git
cd PROJECT
git remote add upstream https://github.com/ORIGINAL-OWNER/PROJECT.git
git remote -v
```

Synchronize your local `main` with the original project:

```bash
git switch main
git fetch upstream
git merge upstream/main
git push origin main
```

This example assumes the original repository's default branch is named `main`. Use the actual branch name if it differs.

## 9.5 Code review basics

A useful review should focus on:
- Correctness and edge cases.
- Readability and maintainability.
- Security and privacy.
- Tests and expected behavior.
- Whether the change matches the task.

Keep feedback specific and respectful. Explain the concern and, where useful, suggest a possible improvement.

## 9.6 Merge options on GitHub

Depending on repository settings, GitHub may offer:
- **Create a merge commit:** preserves the branch history.
- **Squash and merge:** combines the PR's commits into one commit on the base branch.
- **Rebase and merge:** replays commits onto the base branch.

Teams should agree on a preferred strategy.

## 9.7 Keep your branch updated

If `main` changes while you work, you may need to update your feature branch:

```bash
git fetch origin
git switch feature/update-docs
git merge origin/main
```

Alternatively, a team may use rebase:

```bash
git rebase origin/main
```

Do not rebase commits that others depend on unless the team has agreed to it.

## 9.8 GitHub safety and good practices

- Never commit API keys, passwords, tokens, or private keys.
- Use branch protection and required reviews for important branches.
- Read the diff before opening a PR.
- Keep PRs focused and reasonably small.
- Use issues to track bugs and planned work.
- Follow the repository's contribution guidelines and code of conduct.

## Key takeaways

- Branches isolate work; pull requests propose changes for review.
- Forks are useful when contributing without write access.
- `origin` often points to your fork; `upstream` often points to the original project.
- Review, checks, and team conventions help maintain quality.
