# 10. Undoing Changes

Learn how to correct mistakes safely. These commands can discard work, so understand which area they affect before running them.

> **Environment note:** Concepts apply to Windows, macOS, and Linux. Commands are shown in a cross-platform way wherever possible. Screenshot placeholders are intentionally written for **Windows only** (PowerShell, Command Prompt, or Git Bash).

## 10.1 First rule: inspect before changing

Before undoing anything, run:

```bash
git status
git log --oneline --decorate -5
git diff
git diff --staged
```

If the work matters, make a backup or create a temporary branch before attempting recovery.

## 10.2 Discard unstaged changes in a file

```bash
git restore README.md
```

This replaces the working-directory version with the version from the index (usually the last staged version). Uncommitted edits to that file are discarded.

**Warning:** This can permanently discard uncommitted work.

## 10.3 Unstage a staged file

```bash
git restore --staged README.md
```

This removes the file from the staging area but keeps its working-directory changes.

## 10.4 Restore a file from a specific commit

```bash
git restore --source=HEAD~1 -- README.md
```

This places the version of `README.md` from the previous commit into your working directory. It does not move the current branch or automatically commit the result.

Review the result, then stage and commit it if that is what you intend.

## 10.5 Amend the most recent commit

If you forgot to include a file or want to improve the last commit message:

```bash
git add forgotten-file.txt
git commit --amend
```

To change only the message:

```bash
git commit --amend -m "Correct commit message"
```

Amending creates a replacement commit with a new commit ID. Avoid amending commits that have already been shared unless you understand the consequences and coordinate with collaborators.

## 10.6 Reset: move the current branch pointer

`git reset` changes where the current branch points. Its modes differ:

### Soft reset

```bash
git reset --soft HEAD~1
```

Moves the branch back one commit but keeps the changes staged.

### Mixed reset (default)

```bash
git reset HEAD~1
```

Moves the branch back one commit and leaves the changes in the working directory, unstaged.

### Hard reset

```bash
git reset --hard HEAD~1
```

Moves the branch back one commit and resets the index and working directory to match. **This discards tracked uncommitted changes and can make commits harder to recover. Use only when you are certain.**

`HEAD~1` means the first parent of the current commit. Do not run reset commands on shared branches without understanding the impact.

## 10.7 Revert: undo a commit with a new commit

```bash
git revert COMMIT_ID
```

`git revert` creates a new commit that reverses the changes introduced by an earlier commit. It is generally safer than rewriting history on a shared branch.

Find a commit ID with:

```bash
git log --oneline
```

For a merge commit, reverting requires choosing a mainline parent with `-m`; do not guess the parent number.

## 10.8 Stash: temporarily put changes aside

Stash is useful when you need to switch branches but are not ready to commit your current changes.

```bash
git stash
git stash list
git stash apply
```

- `git stash` saves tracked working-tree and staged changes, then cleans them from the working tree.
- `git stash apply` reapplies the stash and keeps it in the stash list.
- `git stash pop` reapplies and removes the stash if successful.

Include untracked files with:

```bash
git stash -u
```

Inspect a stash before applying it:

```bash
git stash show -p stash@{0}
```

## 10.9 Recover a previous branch or commit reference with reflog

Git records recent movements of `HEAD` in the reflog:

```bash
git reflog
```

If you accidentally moved a branch, reflog may help you locate the earlier commit. You can create a recovery branch from a known commit:

```bash
git branch recovery-branch COMMIT_ID
```

Replace `COMMIT_ID` with the correct hash found in the reflog. Reflog entries are local and are not a permanent backup.

## 10.10 Which command should I use?

| Situation | Typical command |
|---|---|
| Discard unstaged file edits | `git restore FILE` |
| Unstage a file but keep edits | `git restore --staged FILE` |
| Replace the last commit with an amended one | `git commit --amend` |
| Move branch back but keep changes staged | `git reset --soft` |
| Move branch back and unstage changes | `git reset` |
| Discard tracked changes and move branch | `git reset --hard` (dangerous) |
| Safely reverse a published commit | `git revert COMMIT_ID` |
| Temporarily save unfinished work | `git stash` |
| Find earlier HEAD positions | `git reflog` |

## 10.11 Safety checklist

Before using a destructive command:
1. Run `git status`.
2. Inspect staged and unstaged changes.
3. Confirm the branch you are on.
4. Make a backup if the work is valuable.
5. Prefer `git revert` for undoing commits already shared with others.
6. Avoid `reset --hard` and force-push unless you understand exactly what will be discarded or rewritten.

## Final takeaway

Git offers several ways to undo mistakes, but they are not interchangeable. Choose based on whether you want to change the working directory, staging area, local history, or shared history.
